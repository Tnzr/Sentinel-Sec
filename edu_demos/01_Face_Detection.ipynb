# Workshop 1 — Face Detection

This notebook demonstrates detecting faces with the project's `FaceDetector` module and visualizing detections inline using `matplotlib`.

**Steps:**
- Import and initialize `FaceDetector`
- Load an image
- Run detection
- Show annotated image and a grid of cropped faces

> Note: This notebook uses `matplotlib` instead of `cv2.imshow` so it works inside Jupyter.

```python
# Ensure project path is importable
import os, sys
proj_root = os.path.abspath(os.path.join('..', 'lib'))
if proj_root not in sys.path:
    sys.path.insert(0, proj_root)

# Imports
import cv2
import numpy as np
import matplotlib.pyplot as plt
from face_detector import FaceDetector, select_image_file

# Create detector
detector = FaceDetector()

# Choose image: either set IMAGE_PATH or use select_image_file() helper
IMAGE_PATH = None  # <-- set to a filepath string to skip dialog
if IMAGE_PATH is None:
    try:
        IMAGE_PATH = select_image_file()
    except Exception:
        pass

if not IMAGE_PATH:
    raise RuntimeError('No image selected. Set IMAGE_PATH to a local image path.')
print(f"Using image: {IMAGE_PATH}")
```

```python
# Load image and run detection
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise RuntimeError(f'Could not read image: {IMAGE_PATH}')
# convert BGR -> RGB for matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
faces = detector.process_frame(img)

print(f"Detected {len(faces)} faces")

# Display the original image before drawing detections
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Annotated image visualization
def show_image(title, image, figsize=(12,8)):
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.title(title)
    plt.axis('off')
    
annot = detector.visualize_detection(img.copy(), faces)
annot_rgb = cv2.cvtColor(annot, cv2.COLOR_BGR2RGB)
show_image('Detections', annot_rgb)
```

```python
# Build a grid of detected faces (cropped)
grid = detector.extract_faces_grid(img, faces, face_size=(140,140))
if grid is None or grid.size == 0:
    print('No faces to show in grid.')
else:
    grid_rgb = cv2.cvtColor(grid, cv2.COLOR_BGR2RGB)
    show_image('Faces Grid', grid_rgb, figsize=(10,6))
```
